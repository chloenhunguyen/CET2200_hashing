##Part 2: Separate Chaining
1. Implement a hash table with separate chaining using **linked lists**.

2. Support the following operations:

`put(key, value)`

`get(key)`

`contains(key)`

`remove(key)`

3. Test the program by inserting at least 10 keys that produce collisions and demonstrate successful retrieval.

```python
# Step 1 - Create a Node

class Node:
  def __init__(self, key, value):
    self.key = key
    self.value = value
    self.next = None

# Step 2 - Create hash table

class ChainedHashTable:
  def __init__(self, table_size):
    self.table_size = table_size
    self.table = [None] * table_size

  def hash_function(self, key):
    return key % self.table_size

# Step 3 - put()

# Logic of put: The logic: Calculate index -> if bucket empty, add node.
# Otherwise, it walk through linked list
# If key already exists, it will replace the value
# If not, it will add new node

  def put(self, key, value):
    index = self.hash_function(key)
    new_node = Node(key, value)

    # Fixed syntax: self.table[index] is None
    if self.table[index] is None:
      self.table[index] = new_node
      return

    current = self.table[index]
    while True:
      if current.key == key:
        current.value = value
        return

      if current.next is None:
        break
      current = current.next

    current.next = new_node

# Step 4 - get()
# Searching the linked list in the bucket

  def get(self, key):
    index = self.hash_function(key)

    current = self.table[index]

    while current is not None:
        if current.key == key:
            return current.value

        current = current.next

    return None

# Step 5 - contains()
  def contains(self, key):
    index = self.hash_function(key)

    current = self.table[index]

    while current is not None:
        if current.key == key:
            return True

        current = current.next

    return False

# Step 6 - remove()

  def remove(self, key):
    index = self.hash_function(key)

    current = self.table[index]
    previous = None

    while current is not None:

        if current.key == key:

            if previous is None:
                self.table[index] = current.next
            else:
                previous.next = current.next

            return True

        previous = current
        current = current.next

    return False

# Create display function

  def display(self):
    for i in range(self.table_size):
      print(f"Bucket {i}: ", end="")
      current = self.table[i]
      nodes = []
      while current:
        nodes.append(f"[{current.key}: {current.value}]")
        current = current.next
      print(" -> ".join(nodes) if nodes else "None")
```

```python
# Testing put() operation
hash_table = ChainedHashTable(10)
keys = [10, 15, 22, 25, 35, 37, 45, 48, 55, 52, 65, 63, 75, 73, 85, 82, 95, 101, 105]

for key in keys:
  hash_table.put(key, f"Value {key}")

hash_table.display()
```

```python
# Testing get() operation
print("Get 55:", hash_table.get(55))
print("Get 95:", hash_table.get(95))

# Testing contains() operation
print("Contains 75:", hash_table.contains(75))
print("Contains 99:", hash_table.contains(99))

# Testing remove() operation
print("\nRemoving 55:")
hash_table.remove(55)

hash_table.display()
```

##Part 3: Open Addressing – Linear Probing
1. Implement the same hash table using open addressing with linear probing.

2. Insert values into the table and handle collisions using linear probing.

3. Write test cases where clustering occurs and show the runtime effect when searching for an element

```python
# Step 1 - Create a Linear Probing Hash Table

class LinearProbingHashTable:
  def __init__(self,table_size):
    self.table_size = table_size
    self.table = [None] * table_size

  def hash_function(self, key):
    return key % self.table_size

# Step 2 - Create put() operation
  def put(self,key,value):
    index = self.hash_function(key)
    for i in range(self.table_size):
      probe_index = (index + i) % self.table_size

      if self.table[probe_index] is None:
        self.table[probe_index] = (key,value)
        return

      if self.table[probe_index][0] == key:
        self.table[probe_index] = (key,value)
        return

# Step 3 - Create get() operation

  def get(self,key):
      index = self.hash_function(key)
      probes = 0

      for i in range(self.table_size):
        probe_index = (index + i) % self.table_size
        probes += 1

        if self.table[probe_index] is None:
          return None, probes

        if self.table[probe_index][0] == key:
          return self.table[probe_index][1], probes

      return None, probes

  # Display operation
  def display(self):
    for i in range(self.table_size):
      if self.table[i] is None:
        print(f"Slot {i}: None")
      else:
        print(f"Slot {i}: [{self.table[i][0]}: {self.table[i][1]}]")

# Testing
linear_table = LinearProbingHashTable(10)
keys = [10, 15, 22, 25, 35, 37, 45, 48, 55, 52, 65, 63, 75, 73, 85, 82, 95, 101, 105]

for key in keys:
  linear_table.put(key, f"Value {key}")

linear_table.display()
```

```python
# Testing runtime effect when searching for an element - Example A

value, probes = linear_table.get(5)

print("Searching for 5:")
print("Value:", value)
print("Probes:", probes)

# Testing runtime effect when searching for an element - Example B

value, probes = linear_table.get(45)

print("\nSearching for 45:")
print("Value:", value)
print("Probes:", probes)
```

**Why?**

Searching for key 5 required only one probe because it was stored at its original hash index. Searching for key 45 required five probes because linear probing created a cluster of occupied positions

##Part 4: Open Addressing – Quadratic Probing
1. Modify the open addressing implementation to use quadratic probing.

2. Test with the same set of inputs as in Part 3, comparing the distribution of keys in the table.

3. Highlight cases where quadratic probing fails to find empty slots even though space is available.

```python
# Step 1 - Create Quadratic Probing Hash Table

class QuadraticProbingHashTable:

    def __init__(self, table_size=10):
        self.table_size = table_size
        self.table = [None] * table_size

    # Hash function
    def hash_function(self, key):
        return key % self.table_size

    # Insert key-value pair
    def put(self, key, value, show_probes=False):
        index = self.hash_function(key)

        if show_probes:
            print(f"\nTrying to insert key {key}")
            print(f"Natural hash index: {key} % {self.table_size} = {index}")

        for i in range(self.table_size):

            # Quadratic probing formula
            probe_index = (index + i**2) % self.table_size

            if show_probes:
                print(f"Probe {i}: checking index {probe_index}")

            # Empty position found
            if self.table[probe_index] is None:
                self.table[probe_index] = (key, value)

                if show_probes:
                    print(f"Key {key} inserted at index {probe_index}")

                return True

            # Key already exists -> update value
            if self.table[probe_index][0] == key:
                self.table[probe_index] = (key, value)
                return True

        # Could not find an empty position
        if show_probes:
            print(f"Failed to insert key {key}")

        return False

    # Search for a key
    def get(self, key):
        index = self.hash_function(key)

        for i in range(self.table_size):
            probe_index = (index + i**2) % self.table_size

            if self.table[probe_index] is None:
                return None

            if self.table[probe_index][0] == key:
                return self.table[probe_index][1]

        return None

    # Display table
    def display(self):
        for i in range(self.table_size):
            print(f"Index {i}: {self.table[i]}")
```

```python
# Testing quadratic probing

quadratic_table = QuadraticProbingHashTable(10)
keys = [10, 15, 22, 25, 35, 37, 45, 48, 55, 52, 65, 63, 75, 73, 85, 82, 95, 101, 105]

for key in keys:
  quadratic_table.put(key, f"Value {key}")

quadratic_table.display()
```

```python
# Example where quadratic probing fails to find empty slots even though the space is available

failure_table = QuadraticProbingHashTable(10)

# All of these keys hash to index 0
keys = [0, 10, 20, 30, 40, 50]

for key in keys:
    failure_table.put(key, f"Value {key}")

print("\nTable before trying to insert 60:")
failure_table.display()
```

```python
print("\nTrying to insert 60:")

success = failure_table.put(
    60,
    "Value 60",
    show_probes=True
)

print("\nInsertion successful:", success)
```

## Part 5: Analysis Questions

1. The difference between separate chaining and open addressing.
* **Separate chaining**: keeps colliding values in the same bucket. Each bucket can contain a linked list of multiple key-value pairs.
* **Open addressing**: stores every key value pair directly inside the array. When a collision occurs, the algorithm searches other array positions using a probing method


2. What causes primary clustering in linear probing, and how quadratic probing tries to fix it.

Linear probing creates primary clustering because colliding values are stored consecutively, producing long groups of occupied cells.
Quadratic probing reduces this problem by jumping farther from the original hash index using squared offsets rather than checking only adjacent positions

3. Why choosing a prime table size is important for quadratic probing.

A prime table size is useful in quadratic probing because it reduces repeated probe positions and provides a stronger guarantee that many different table locations will be checked. This reduces the chance of repeatedly cycling through the same indices while unused positions still exist.

4. Given a sequence of insertions, identify the load factor and decide when resizing should occur.

* The formula: load factor = number of items / table capacity -> λ = n / c

The load factor measures how full the hash table is. It is calculated as the number of stored elements divided by the table capacity. For example, if 6 items are stored in a table of size 10, the load factor is 0.6. As the load factor increases, collisions and probing become more frequent. For open addressing, resizing around a load factor of 0.5 is a useful rule of thumb.
